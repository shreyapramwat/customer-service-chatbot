```python
import pandas as pd
import numpy as np

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

import re
```

```python
faq_data = {
    "questions": [
        "hello",
        "hi",
        "hey",
        "good morning",
        "how are you",

        "where is my order",
        "track my order",
        "how can I track my package",
        "when will my order arrive",

        "what payment methods are accepted",
        "can I pay using credit card",
        "do you accept online payments",

        "how can I return a product",
        "what is your return policy",
        "can I exchange my product",

        "how can I get a refund",
        "when will I receive my refund",
        "my refund is not received",

        "how can I contact customer support",
        "what is your contact number",
        "how do I contact you",

        "thank you",
        "thanks",
        "bye",
        "goodbye"
    ],

    "answers": [
        "Hello! Welcome to our customer service. How can I help you?",
        "Hello! Welcome to our customer service. How can I help you?",
        "Hi there! How may I assist you?",
        "Good morning! How can I help you today?",
        "I'm doing great! How can I assist you?",

        "You can track your order using the tracking link sent to your registered email.",
        "You can track your order using the tracking link sent to your registered email.",
        "Please check your registered email for your package tracking link.",
        "Delivery usually takes 3-7 business days. Check your order tracking for updates.",

        "We accept credit cards, debit cards, UPI and net banking.",
        "Yes! We accept major credit and debit cards.",
        "Yes! We support online payments through multiple payment methods.",

        "You can return eligible products within 7 days of delivery.",
        "Our return policy allows eligible products to be returned within 7 days.",
        "Yes! Product exchanges are available for eligible items.",

        "You can request a refund through the order section of your account.",
        "Refunds are generally processed within 5-7 business days after approval.",
        "Please contact our support team with your order ID to check your refund status.",

        "You can contact our support team at support@example.com.",
        "Our customer support email is support@example.com.",
        "You can email us at support@example.com for assistance.",

        "You're welcome! Happy to help.",
        "You're welcome! Let us know if you need anything else.",
        "Goodbye! Have a wonderful day.",
        "Thank you for contacting us. Goodbye!"
    ]
}

print("FAQ dataset created successfully!")
```

```python
questions = faq_data["questions"]
answers = faq_data["answers"]

# Create TF-IDF vectorizer
vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2)
)

# Convert questions into numerical vectors
X = vectorizer.fit_transform(questions)

print("NLP model prepared successfully!")
print("Total questions:", len(questions))
```

```python
def chatbot_response(user_input):

    # Convert user input into a vector
    user_vector = vectorizer.transform([user_input])

    # Calculate similarity
    similarity_scores = cosine_similarity(
        user_vector, X
    )

    # Find the most similar question
    best_match = np.argmax(similarity_scores)

    # Get similarity score
    confidence = similarity_scores[0][best_match]

    # Set minimum confidence threshold
    if confidence < 0.20:
        return "Sorry, I couldn't understand your question. Please contact our customer support team."

    # Return matching answer
    return answers[best_match]
```

```python
test_questions = [
    "Hello",
    "Where can I track my package?",
    "Do you accept UPI?",
    "I want to return my order",
    "How long does a refund take?",
    "How can I contact support?"
]

for question in test_questions:

    response = chatbot_response(question)

    print("Customer:", question)
    print("Chatbot:", response)
    print("-" * 60)
```

```python
print("===================================")
print("   CUSTOMER SERVICE CHATBOT")
print("===================================")

print("Welcome! How can I help you today?")
print("Type 'exit' to end the conversation.")

while True:

    user_input = input("\nYou: ")

    if user_input.lower().strip() == "exit":
        print("Bot: Thank you for contacting us. Goodbye!")
        break

    response = chatbot_response(user_input)

    print("Bot:", response)
```

```python
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

chat_output = widgets.Output()

user_input = widgets.Text(
    placeholder="Type your question here...",
    description="You:",
    layout=widgets.Layout(width="80%")
)

send_button = widgets.Button(
    description="Send",
    button_style="success"
)

def send_message(button):

    message = user_input.value.strip()

    if not message:
        return

    with chat_output:
        print("You:", message)

        if message.lower() == "exit":
            print("Bot: Thank you! Goodbye!")
        else:
            print("Bot:", chatbot_response(message))

        print("-" * 40)

    user_input.value = ""

send_button.on_click(send_message)

display(HTML("<h2>Customer Service Chatbot</h2>"))
display(chat_output)
display(widgets.HBox([user_input, send_button]))
```
